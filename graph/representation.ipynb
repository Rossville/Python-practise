<p>Representation of graph in <strong>Python</strong></p>
<ol>
    <li>Adjacency Matrix</li>
    <li>Adjacency List</li>
</ol>

<p style="font-size: 25px;">1. Adjacency Matrix (for undirected graph)</p>

```python
from typing import List

def addEdge(mat: List[List[int]], i: int, j: int) -> None:
    mat[i][j] = 1
    mat[j][i] = 1
    # for directed graphs, remove mat[j][i] = 1 (only keep 1 statement and not both)
    
def displayMatrix(mat) -> None:
    for i in range(len(mat)):
        print(i,end="")
        for j, val in enumerate(mat[i]):
            if val == 1:
                print(f"-> {j}",end="")
        print()

if __name__ == "__main__":
    V = 5
    mat = [[0]*V for _ in range(V)]
    addEdge(mat, 0, 1)
    addEdge(mat, 0, 3)
    addEdge(mat, 0, 4)
    addEdge(mat, 1, 2)
    addEdge(mat, 3, 2)
    addEdge(mat, 4, 3)
    displayMatrix(mat)
```

<p style="font-size: 25px;">2. Adjacency List (for undirected graph)</p>

```python
def addEdge(adj_list: List[List[int]], i: int, j: int) -> None:
    # for undirected graph
    adj_list[i].append(j)
    # adj_list[j].append(i)
    # for directed graph, remove this line - adj_list[j].append(i) 

def display_list(adj_list: List[List[int]]) -> None:
    for ind,i in enumerate(adj_list):
        if len(i) == 0:
            continue
        print(ind,end="")
        for j in i:
            print(f" -> {j}",end="")
        print()

if __name__ == "__main__":
    V = 5
    adj_list = [[] for _ in range(V)]
    addEdge(adj_list, 0, 1)
    addEdge(adj_list, 0, 3)
    addEdge(adj_list, 0, 4)
    addEdge(adj_list, 1, 2)
    addEdge(adj_list, 3, 2)
    addEdge(adj_list, 4, 3)
    display_list(adj_list)
```
