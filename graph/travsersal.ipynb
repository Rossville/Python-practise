<p style="font-size: 25px;">Breadth First Search on a Graph</p>

```python
from typing import Deque, List
from collections import deque

def BFS(adj_list: List[List[int]]) -> List[int]:
    visited: List[int] = [0] * len(adj_list)
    queue: deque = deque()
    queue.append(1)
    visited[1] = 1
    bfs_order = []
    
    while queue:
        node = queue.popleft()
        bfs_order.append(node)
        for neighbor in adj_list[node]:
            if visited[neighbor] == 0:
                visited[neighbor] = 1
                queue.append(neighbor)
    
    return bfs_order
    
def addEdge(adj_list: List[List[int]], i: int, j: int) -> None:
    adj_list[i].append(j)
    adj_list[j].append(i)

def display(adj_list: List[List[int]]) -> None:
    for indx, i in enumerate(adj_list):
        if len(i) == 0:
            continue
        print(indx,end="")
        for j in i:
            print(f" -> {j}", end="")
        print()

if __name__ == "__main__":
    V=9
    adj_list: List[List[int]] = [[] for _ in range(V+1)]
    addEdge(adj_list, 1, 2)
    addEdge(adj_list, 1, 6)
    addEdge(adj_list, 2, 3)
    addEdge(adj_list, 2, 4)
    addEdge(adj_list, 6, 7)
    addEdge(adj_list, 6, 9)
    addEdge(adj_list, 4, 5)
    addEdge(adj_list, 7, 8)
    addEdge(adj_list, 5, 8)
    display(adj_list)
    print()
    vis: List[int] = BFS(adj_list)
    print(vis)
```
