<p>Read from the <strong>data.json</strong> and print the output in the structured format.</p>

```python
import json
from typing import Dict, List

def read_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
    return data

def display_user(data) -> None:
    print("Display the users data with their connections\n")
    for user in data["users"]:
        print(f"{user["name"]} (ID: {user['id']}) friends: {user["friends"]}")
    
display_user(read_data("data.json"))
    
    
```
