<p><strong>Series</strong> is like a dict, they can store and access data similar to dict</p>

```python
import pandas as pd
import numpy as np

data = {"a":1, "b":2, "c":3}
s = pd.Series(data)
try:
    s["d"] = 4
    s["e"] = 5

    s["f"] # this will raise an exception
    s.get("f") # this and the below one won't raise an exception instead it'll return None and below one will return 'nan' if "f" does not exist on the Series 's'
    s.get("f", np.nan)
except KeyError as e:
    print(f"Key does not exits: {e}")
```

<b>Vectorized operations and label alignment with Series</b>
<p style="font-size:15px">When working with Series, looping through each element in the series is not necessary</p>


```python
import pandas as pf

s1 = pf.Series([[1,2,3], [4,5,6], [7,8,9]])
s2 = pf.Series([[1,2,3], [4,5,6], []])
a = pf.Series([1,2,3])
b = pf.Series([4,5,6])

x = s1 + s2 # concatenation
y = pf.Series(a * b)
y.head()
```
