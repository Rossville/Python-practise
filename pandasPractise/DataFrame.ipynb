<b style="font-size: 25px">DataFrame in pandas</b>

<p>1. From dicts of Series</p>

```python
import pandas as pd

data = {
    "two" : pd.Series([3,4,2], index=["a","b","c"]),
    "one" : pd.Series([1,2,3,4], index=["a","b","c","d"]),
}

df = pd.DataFrame(data)
df.head()
```

<p>2. Taking Input from <strong>CSV</strong> File</p>

```python

df = pd.read_csv('./studentMarks.csv')

# Get the data type of the object
df.dtypes

# Printing the data
df.head()

# Mean of all columns
df.mean()

# Median of all columns
df.median()

df.describe()
```

<p>3. Taking input from dict of <strong>ndarrays/list</strong></p>
<describe>All ndarrays must share same length and the index passed should also be of same length, If no index is passed then range(n) will be taken for index array as default, else following exception will be raised</describe>
<p style="font-weight: bold;">ValueError: All arrays must be of the same length</p>

```python

d = {"one": [1,2,3,4], "two": [5,10,11,12]}
df = pd.DataFrame(d, index=["a","b","c","d"])
df.head()
```

<p>4. Taking Input from <strong>Structured and record</strong> array</p>

```python
import numpy as np

d = np.zeros((2,), dtype = [("A","i4"),("B","f4"),("C","S10")]) # s10 is a numpy byte string of 10bytes, to use normal python string use U10 (Unicode 10byte string)

d[:] = [(1, 2, "Hello"),(3, 4, "World")]

df = pd.DataFrame(d)
df.head()

```
