<p>Series can be instantiated as dict</p>

```python
import pandas as pd

data = {"b":1, "c":2, "d":3}
s = pd.Series(data)
s.head()
```

<p>Since no matching <strong>keys</strong> were found hence <strong>NaN</strong> is displayed</p>

```python
import pandas as pd

data = {"b":1, "c":2, "d":3}
s = pd.Series(data, index=["x","y","z","m"])
s.head()
```

<p><strong>Series</strong> is like a dict, they can store and access data similar to dict</p>

```python
import pandas as pd
import numpy as np

data = {"a":1, "b":2, "c":3}
s = pd.Series(data)
try:
    s["d"] = 4
    s["e"] = 5

    s["f"] # this will raise an exception
    s.get("f") # this and the below one won't raise an exception instead it'll return None and below one will return 'nan' if "f" does not exist on the Series 's'
    s.get("f", np.nan)
except:
    print(f"Following exception is raised:")
```
