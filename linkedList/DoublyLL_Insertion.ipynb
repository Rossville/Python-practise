<b style="font-size: 25px;">Doubly Linked List</b>

<p>&nbsp;&nbsp;&nbsp;&nbsp;Initialzing a Doubly Linked List</p>

```python
from typing import Optional, List
class ListNode:
    def __init__(self, val:int=0, next: Optional['ListNode']=None, prev: Optional['ListNode']=None) -> None:
        self.val = val
        self.next = next
        self.prev = prev

def createDLL() -> ListNode:
    listVal: List[int] = [1,2,3,4,5,6,7,8]
    head: ListNode = ListNode(listVal[0])
    curr: ListNode = head
    for index, val in enumerate(listVal):
        newNode: ListNode = ListNode(val)
        if index > 0:
            curr.next = newNode
            newNode.prev = curr
            curr = curr.next
    return head

def printDLL(headNode: ListNode | None) -> None:
    curr: ListNode | None = headNode
    print("Printing the Doubly Linked List(from start to end):")
    while curr is not None:
        print(curr.val, end='')
        if curr.next is not None:
            print(' -> ', end='')
        curr = curr.next

def printDLL_end_to_start(headNode: ListNode | None) -> None:
    curr: ListNode | None = headNode
    print("Printing the Doubly Linked List(from end to start):")
    while curr.next is not None:
        if curr is not None:
            curr = curr.next
    # now 'curr' represent the ending node of the doubly linked list.
    while curr is not None:
        print(curr.val,end='')
        if curr.prev is not None:
            print(' <- ',end='')
        curr = curr.prev
    
def main() -> None:
    printDLL(createDLL())
    print()
    printDLL_end_to_start(createDLL())

if __name__ == "__main__":
    main()
```

<b style="font-size: 25px;">Insertion in the Doubly Linked List</b>

<p>&nbsp;&nbsp;&nbsp;&nbsp;1. Insertion at head in the doubly linked list</p>

```python
def insertionAtHead(val: int, headNode: ListNode | None) -> ListNode | None:
    newNode: ListNode = ListNode(val)
    if headNode is None:
        return newNode
    newNode.next = headNode
    headNode.prev = newNode
    headNode = newNode
    return newNode

def main() -> None:
    dll: ListNode = createDLL()
    printDLL(dll)
    print()
    newHead: ListNode | None = insertionAtHead(100, dll)
    printDLL(newHead)
    
if __name__ == "__main__":
    main()    
    
```

<p>&nbsp;&nbsp;&nbsp;&nbsp;2. Insertion at the end of the doubly linked list</p>

```python
def insertionAtTail(val:int,headNode: ListNode | None) -> None | ListNode:
    newNode: ListNode = ListNode(val)
    if headNode is None:
        return newNode
    curr: ListNode = headNode
    while curr.next is not None:
        curr = curr.next
    curr.next = newNode
    
def main() -> None:
    dll: ListNode = createDLL()        
    insertionAtTail(800, dll)
    printDLL(dll)
main()
```

<p>&nbsp;&nbsp;&nbsp;&nbsp;3. Insertion at an index 'k' from the head in the doubly linked list</p>
<p>&nbsp;&nbsp;&nbsp;&nbsp;'k' follow 0-based indexing</p>

```python
def insertionAtK(k: int, val: int, headNode: ListNode | None) -> ListNode | None:
    newNode: ListNode | None = ListNode(val)
    curr: ListNode | None = headNode
    i: int = 1
    if headNode is None:
        return newNode
    if k == 0:
        newNode.next = headNode
        headNode.prev = newNode
        return newNode
    while i!=k :
        i+=1
        if curr.next is not None:
            curr = curr.next
        else:
            break
    if curr.next is not None:
        newNode.next = curr.next
        curr.next = newNode
        newNode.prev = curr
        curr.next.next.prev = newNode
        return headNode
    curr.next = newNode # if k is greater than the lenght of linked list, then we'll add the element at the end or if the k represents the end element then also we'll add at the end.
    return headNode
    
def main() -> None:
    dll: ListNode | None = createDLL()
    x = insertionAtK(1, 500, dll)
    printDLL(x)

if __name__ == "__main__":
    main()
        
```
