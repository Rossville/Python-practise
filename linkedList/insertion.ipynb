<b style="font-size: 25px;">Initializing a Linked List(Singly Linked List)</b>

```python
class Node:
    # Integer Linked List
    def __init__(self, data: int) -> None:
        self.data = data
        self.next = None

def createLinkedList() -> Node:
    node1: Node = Node(5)
    node2: Node = Node(4)
    node3: Node = Node(3)
    node4: Node = Node(2)
    node5: Node = Node(1)
        
    node1.next = node2
    node2.next = node3
    node3.next = node4
    node4.next = node5
        
    head = node1
    return head

def printLinkedList(headNode: Node) -> None:
    curr: Node = headNode
    while curr.next != None:
        print(f"{curr.data} -> ", end="")
        curr = curr.next
    print(curr.data) # last element of the linked list.

def main() -> None:
    printLinkedList(createLinkedList())
    
if __name__ == "__main__":
    main()
```

<b style="font-size: 25px;">Inserting an element into the linked list</b>

<p>&nbsp;&nbsp;&nbsp;&nbsp;1. Inserting the element at the begining of the linked list</p>

```python
def main() -> None:
    headNode: Node = createLinkedList()
    newStartNode: Node = Node(100) # 100 will be the new starting node of the list.
    newStartNode.next = headNode
    print("Insertion at the begining of the linked list: ", end="")
    printLinkedList(newStartNode)
    
if __name__ == "__main__":
    main()
```

<p>&nbsp;&nbsp;&nbsp;&nbsp;2. Inserting at the end of the linked list</p>

```python
def main() -> None:
    headNode: Node = createLinkedList()
    newEndNode: Node = Node(120) # 120 will be the new end of the given linked list.
    print("Linked List before insertion at the end: ", end="")
    printLinkedList(headNode)
    def endNode(headNode: Node) -> Node:
        curr: Node = headNode
        while curr.next != None:
            curr = curr.next
        return curr
    endNode(headNode).next = newEndNode
    print("Linked list after insertion at the end: ", end="")
    printLinkedList(headNode)

if __name__ == "__main__":
    main()
            
```

<p>&nbsp;&nbsp;&nbsp;&nbsp;3. Inserting at a given position in the linked list</p>

```python
def main() -> None:
    post: int = 3
    ElementToInsert: Node = Node(800) # 800 will be inserted at 3 position of the linked list.
    headNode: Node = createLinkedList()
    
    print(f"Linked list before insertion at {post} : ",end="")
    printLinkedList(headNode)
    
    def InsertAtPosition(post: int, ElementToInsert: Node, headNode: Node) -> None:
        curr: Node = headNode
        i: int = 0
        beforeNodeElement: Node = Node(0)
        while i < post and curr.next != None:
            if i == post-1:
                beforeNodeElement = curr
            curr = curr.next
            i += 1
        beforeNodeElement.next = ElementToInsert
        ElementToInsert.next = curr
    
    InsertAtPosition(post, ElementToInsert, headNode)
    print(f"Linked list after insertion at {post}: ", end="")
    printLinkedList(headNode)
    
main()
        
```
