<b style="font-size: 25px;">Deletion in Linked List(Singly Linked List)</b>

<p>Node Initialization :</p>

```python
from typing import Optional

class Node:
    def __init__(self, val: int = 0, next = None) -> None:
        self.val = val
        self.next = next
    
```

```python
def createLinkedList() -> Node:
    node1: Node = Node(4)
    node2: Node = Node(1)
    node3: Node = Node(7)
    node4: Node = Node(0)
    node5: Node = Node(9)
    
    node1.next = node2
    node2.next = node3
    node3.next = node4
    node4.next = node5
    
    return node1
```

```python
def printList(head: Node | None) -> None:
    curr: Node | None = head
    while curr is not None:
        print(f"{curr.val}", end='')
        curr = curr.next
        if curr is not None:
            print(" -> ", end='')
    
```

```python
def main() -> None:
    ll: Optional['Node'] = createLinkedList()
    printList(ll)
main()
```

<b style="font-size: 25px;">Deletion of Node in the Linked List</b>

<p>&nbsp;&nbsp;&nbsp;&nbsp;1. Deletion at head</p>

```python
def deletionAtHead(headNode: Node | None) -> Node | None:
    if headNode is None:
        return None
    new_head: Node | None = headNode.next
    return new_head

def main2() -> None:
    ll: Node | None = createLinkedList()
    ll = deletionAtHead(ll)
    printList(ll)
    
main2()    
```

<p>&nbsp;&nbsp;&nbsp;&nbsp;2. Deletion at the end of Linked List</p>

```python
def deletionAtEnd(headNode: Node) -> Node | None:
    curr: Node = headNode
    while curr.next.next is not None:
        curr = curr.next
    temp: Node = curr.next
    curr.next = None
    return temp

def main3() -> None:
    ll: Node | None = createLinkedList()
    x: Node | None = deletionAtEnd(ll)
    if x:
        print(f"Last Deleted node : {x.val}")
    printList(ll)
    
main3()   
```

<p>&nbsp;&nbsp;&nbsp;&nbsp;3. Deletion at an index k</p>
<desc>&nbsp;&nbsp;&nbsp;&nbsp;An integer k will be given to the deletionAtk() method, where 'k' represent index. So delete the value present at index k in the linked list.</desc>
<p>&nbsp;&nbsp;&nbsp;&nbsp;'k' is 0-based indexed</p>

```python
def deletionAtK(headNode: Node | None, k: int) -> Node | None:
    i: int = 1
    curr: Node | None = headNode
    # empty linked list
    if headNode is None:
        return None
    if k == 0:
        # delete first node, store and return it
        new_head: Node = headNode.next
        headNode.next = None
        return new_head
    while curr.next.next is not None:
        if i == k:
            temp: Node = curr.next
            curr.next = curr.next.next
            temp.next = None
            return temp
        i += 1
        curr = curr.next
    # check the last element for deletion
    if i == k:
        temp: Node = curr.next
        curr.next = None
        return temp
    
def main4() -> None:
    ll: Node | Node = createLinkedList()
    x: Node = deletionAtK(ll, 0)    # del fist element
    # x: Node = deletionAtK(ll, 2)    # del element at index 2
    # x: Node = deletionAtK(ll, 3)    # del second last
    # x: Node = deletionAtK(ll, 4)    # del last
    printList(x)
    
main4()
```
