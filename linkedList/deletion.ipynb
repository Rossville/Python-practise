<p>Node Initialization :</p>

```python
from typing import Optional

class Node:
    def __init__(self, val: int = 0, next: Optional['Node'] = None) -> None:
        self.val = val
        self.next: Optional['Node'] = next
    
```

```python
def createLinkedList() -> Node:
    node1: Node = Node(4)
    node2: Node = Node(1)
    node3: Node = Node(7)
    node4: Node = Node(0)
    node5: Node = Node(9)
    
    node1.next = node2
    node2.next = node3
    node3.next = node4
    node4.next = node5
    
    return node1
```

```python
def printList(head: Node) -> None:
    curr: Node | None = head
    while curr is not None:
        print(f"{curr.val}", end='')
        curr = curr.next
        if curr is not None:
            print(" -> ", end='')
    
```

```python
def main() -> None:
    ll: Optional['Node'] = createLinkedList()
    printList(ll)
main()
```

<b style="font-size: 25px;">Deletion of Node in the Linked List</b>

<p>&nbsp;&nbsp;&nbsp;&nbsp;1. Deletion at head</p>

```python
def deletionAtHead(headNode: Node | None) -> Node | None:
    if headNode is None:
        return None
    new_head: Node | None = headNode.next
    return new_head

def main2() -> None:
    ll: Node = createLinkedList()
    ll = deletionAtHead(ll)
    printList(ll)
    
main2()    
```
